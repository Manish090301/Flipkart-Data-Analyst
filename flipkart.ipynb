```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv("D:/data analytics/flipkart_cleaned.csv")

# Data Cleaning
df['retail_price'] = pd.to_numeric(df['retail_price'], errors='coerce')
df['discounted_price'] = pd.to_numeric(df['discounted_price'], errors='coerce')

df.dropna(inplace=True)

# Create Discount Column
df['discount'] = df['retail_price'] - df['discounted_price']


# ==============================
# Dashboard 1: Top Categories
# ==============================

plt.figure(figsize=(10,5))

top_category = df['product_category_tree'].value_counts().head(10)

sns.barplot(x=top_category.values, y=top_category.index)

plt.title("Dashboard 1: Top Product Categories")

plt.show()



# ==============================
# Dashboard 2: Top Brands
# ==============================

plt.figure(figsize=(10,5))

top_brands = df['brand'].value_counts().head(10)

sns.barplot(x=top_brands.values, y=top_brands.index)

plt.title("Dashboard 2: Top Brands")

plt.show()



# ==============================
# Dashboard 3: Retail Price Distribution
# ==============================

plt.figure(figsize=(10,5))

sns.histplot(df['retail_price'], bins=50)

plt.title("Dashboard 3: Retail Price Distribution")

plt.show()



# ==============================
# Dashboard 4: Discount Distribution
# ==============================

plt.figure(figsize=(10,5))

sns.histplot(df['discount'], bins=50, color='green')

plt.title("Dashboard 4: Discount Distribution")

plt.show()



# ==============================
# Dashboard 5: Retail vs Discount Price
# ==============================

plt.figure(figsize=(10,5))

sns.scatterplot(x='retail_price', y='discounted_price', data=df)

plt.title("Dashboard 5: Retail vs Discount Price")

plt.show()

```
